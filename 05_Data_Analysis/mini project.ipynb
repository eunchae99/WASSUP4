### 분석1) 학습 패턴


````python
import pandas as pd
df = pd.read_csv('./data/mini.csv')
df
````

````python
# 'target' 컬럼의 데이터 타입 확인
print(df['target'].dtype)
````

````python
# 구독 연장 (target == 1)과 취소 (target == 0) 사용자 데이터 분리
renewed_users = df[df['target'] == 1]
canceled_users = df[df['target'] == 0]
````

````python
# 구독 연장과 취소 그룹의 주요 변수 통계 비교
renewed_stats = renewed_users[['subscription_duration', 'recent_login_time', 'average_login_time', 
                               'average_time_per_learning_session', 'monthly_active_learning_days', 
                               'total_completed_courses', 'recent_learning_achievement', 
                               'abandoned_learning_sessions', 'community_engagement_level', 
                               'payment_pattern', 'target']].describe()

canceled_stats = canceled_users[['subscription_duration', 'recent_login_time', 'average_login_time', 
                                 'average_time_per_learning_session', 'monthly_active_learning_days', 
                                 'total_completed_courses', 'recent_learning_achievement', 
                                 'abandoned_learning_sessions', 'community_engagement_level', 
                                 'payment_pattern', 'target']].describe()

print("Renewed Users Statistics: ", renewed_stats)
print("Canceled Users Statistics: ", canceled_stats)
````

````python
from scipy import stats

# 각 학습 세션에 소요된 평균 시간의 차이 검정 (t-test)
t_stat, p_value = stats.ttest_ind(renewed_users['average_time_per_learning_session'], 
                                  canceled_users['average_time_per_learning_session'])
print(f"T-statistic: {t_stat}, P-value: {p_value}")
````

````python
# 구독 연장 그룹과 구독 취소 그룹으로 나누기
renewed_users = df[df['target'] == 1]
canceled_users = df[df['target'] == 0]

# 각 그룹의 각 학습 세션에 소요된 평균 시간 계산
average_time_renewed = renewed_users['average_time_per_learning_session'].mean()
average_time_canceled = canceled_users['average_time_per_learning_session'].mean()

print(f"구독 연장 그룹의 평균 학습 세션 시간: {average_time_renewed:.2f} 분")
print(f"구독 취소 그룹의 평균 학습 세션 시간: {average_time_canceled:.2f} 분")
````

````python
#average_time_per_learning_session: 각 학습 세션에 소요된 평균 시간 (분)
#구독연장(1): 평균 60.11분
#구독취소(0): 평균 46.46분
#차이: 구독연장 그룹이 구독취소 그룹보다 각 학습 세션당 평균 시간이 더 길게 나타남. 이는 학습에 더 많은 시간을 할애하는 경향이 있음을 증명함.

#학습 세션당 평균 시간을 늘리기 위한 전략!!!
#더 긴 학습 세션을 유도하는 방법(콘텐츠)을 통해 사용자들이 더 오래 학습하게 만들면 구독 연장 가능성이 높아질 수 있음.
````

````python
# 구독 연장자와 취소자의 커뮤니티 참여도 비교
renewed_engagement = df_renewed['community_engagement_level']
canceled_engagement = df_canceled['community_engagement_level']

# 두 그룹의 커뮤니티 참여도 차이 분석
t_statistic, p_value = stats.ttest_ind(renewed_engagement, canceled_engagement)
print(f"T-statistic: {t_statistic}, P-value: {p_value}")
````

````python
# 구독 취소 그룹과 구독 유지 그룹 간에 중단된 학습 세션 수에 유의미한 차이가 있음.
````

````python
import pandas as pd
df = pd.read_csv('./data/mini.csv')
df
````

````python
A = df[['user_id','abandoned_learning_sessions','target']]
A
````

````python
#abandoned_learning_sessions의 값에 따라 그룹화하고 각 그룹에서 target 값의 빈도 계산
grouped = A.groupby('abandoned_learning_sessions')['target'].value_counts().unstack().fillna(0)

#각 그룹에서 '0'과 '1'의 개수 계산
grouped['total'] = grouped[0] + grouped[1]  # 총 개수
grouped['count_of_0'] = grouped[0]  # '0'인 개수
grouped['count_of_1'] = grouped[1]  # '1'인 개수

grouped[['count_of_0', 'count_of_1']]
````

````python
#abandoned_learning_sessions에 대해 0과 1의 비율 계산
grouped['ratio_of_0'] = grouped['count_of_0'] / grouped[['count_of_0', 'count_of_1']].sum(axis=1)
grouped['ratio_of_1'] = grouped['count_of_1'] / grouped[['count_of_0', 'count_of_1']].sum(axis=1)

print(grouped[['ratio_of_0', 'ratio_of_1']])
````

````python
import matplotlib.pyplot as plt

# 각 abandoned_learning_sessions에 대해 0과 1의 비율 계산
grouped['ratio_of_0'] = grouped['count_of_0'] / grouped[['count_of_0', 'count_of_1']].sum(axis=1)
grouped['ratio_of_1'] = grouped['count_of_1'] / grouped[['count_of_0', 'count_of_1']].sum(axis=1)

# 그래프 시각화 (막대그래프)
fig, ax = plt.subplots(figsize=(10, 6))

# 막대그래프 그리기
ax.bar(grouped.index - 0.2, grouped['ratio_of_0'], width=0.4, label='구독 취소 비율', color='blue')
ax.bar(grouped.index + 0.2, grouped['ratio_of_1'], width=0.4, label='구독 연장 비율', color='red')

# 그래프 제목 및 레이블 추가
ax.set_title('중단된 학습 세션 수에 따른 구독연장 여부의 비율')
ax.set_xlabel('중단된 학습 세션 수')
ax.set_ylabel('비율')

# x축 눈금 설정
ax.set_xticks(grouped.index)
ax.set_xticklabels(grouped.index)

# 범례 추가
ax.legend()

# 그래프 출력
plt.tight_layout()
plt.show()
````

````python
# 'marketing'이 True인 경우에 대해 'managed'가 True인 경우와 False인 경우를 구하기
marketing_true = df3[df3['marketing'] == True]

# 'marketing'이 True인 데이터에서 'managed'가 True인 경우
marketing_true_managed_true = marketing_true[marketing_true['managed'] == True]

# 'marketing'이 True인 데이터에서 'managed'가 False인 경우
marketing_true_managed_false = marketing_true[marketing_true['managed'] == False]

# 각각의 개수 출력
print("Marketing이 True인 데이터에서 Managed가 True인 경우 개수:", len(marketing_true_managed_true))
print("Marketing이 True인 데이터에서 Managed가 False인 경우 개수:", len(marketing_true_managed_false))
````

````python
# 'marketing'이 False인 경우에 대해 'managed'가 True인 경우와 False인 경우를 구하기
marketing_false = df3[df3['marketing'] == False]

# 'marketing'이 False인 데이터에서 'managed'가 True인 경우
marketing_false_managed_true = marketing_false[marketing_false['managed'] == True]

# 'marketing'이 False인 데이터에서 'managed'가 False인 경우
marketing_false_managed_false = marketing_false[marketing_false['managed'] == False]

# 각각의 개수 출력
print("Marketing이 False인 데이터에서 Managed가 True인 경우 개수:", len(marketing_false_managed_true))
print("Marketing이 False인 데이터에서 Managed가 False인 경우 개수:", len(marketing_false_managed_false))
````

````python
# 총 Marketing이 True인 경우와 False인 경우의 데이터 개수
total_marketing_true = len(df3[df3['marketing'] == True])
total_marketing_false = len(df3[df3['marketing'] == False])

# 비율 계산
marketing_true_managed_true_ratio = len(marketing_true_managed_true) / total_marketing_true
marketing_true_managed_false_ratio = len(marketing_true_managed_false) / total_marketing_true
marketing_false_managed_true_ratio = len(marketing_false_managed_true) / total_marketing_false
marketing_false_managed_false_ratio = len(marketing_false_managed_false) / total_marketing_false

# 출력
print("Marketing이 True인 데이터에서 Managed가 True인 경우 비율:", marketing_true_managed_true_ratio)
print("Marketing이 True인 데이터에서 Managed가 False인 경우 비율:", marketing_true_managed_false_ratio)
print("Marketing이 False인 데이터에서 Managed가 True인 경우 비율:", marketing_false_managed_true_ratio)
print("Marketing이 False인 데이터에서 Managed가 False인 경우 비율:", marketing_false_managed_false_ratio)
````
